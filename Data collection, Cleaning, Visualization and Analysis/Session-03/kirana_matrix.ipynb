This notebook covers matrix operations, linear algebra, array manipulation and statistical analysis using NumPy.

```python
import numpy as np

print("NumPy Version:", np.__version__)

# Generate Session 01 sales data
np.random.seed(42)
units_sold = np.random.randint(5, 60, size=(6, 7))
price_inr = np.array([52, 44, 45, 380, 140, 120])

print("\nUnits Sold:")
print(units_sold)
print("\nPrice INR:")
print(price_inr)
```

## Task 1: Brand Switching Matrix and Linear Algebra

Create a 3x3 transition matrix for three tea brands. Calculate its determinant, inverse, eigenvalues and normalized steady-state market share.

```python
switch_p = np.array([
    [0.7, 0.2, 0.1],
    [0.3, 0.5, 0.2],
    [0.2, 0.3, 0.5]
])

print("Brand Switching Matrix:")
print(switch_p)

print("\nDeterminant:")
print(np.linalg.det(switch_p))

print("\nInverse:")
print(np.linalg.inv(switch_p))

# Eigenvalues and eigenvectors of transpose
eigenvalues, eigenvectors = np.linalg.eig(switch_p.T)

# Find eigenvector corresponding to eigenvalue 1
index = np.argmin(np.abs(eigenvalues - 1))
steady_share = np.real(eigenvectors[:, index])

# Normalize eigenvector
steady_share = steady_share / steady_share.sum()

print("\nEigenvalues:")
print(eigenvalues)

print("\nSteady-State Market Share:")
print(steady_share)

print("\nVerification:", np.allclose(switch_p.T @ steady_share, steady_share))
print("Total Market Share:", steady_share.sum())
```

## Task 2: Array Manipulation and Memory Sharing

Generate a second week, concatenate and stack the two weeks, split them back into their original arrays, and compare flatten() with ravel().

```python
# Preserve first week sales
week1 = units_sold.copy()

# Generate second week
week2 = np.random.randint(5, 60, size=(6, 7))

# Concatenate along columns
two_weeks = np.concatenate((week1, week2), axis=1)

print("Two Weeks Shape:", two_weeks.shape)
print(two_weeks)

# Stack along a new axis
stacked = np.stack((week1, week2), axis=0)

print("\nStacked Shape:", stacked.shape)
print(stacked)

# Split combined array into two weeks
split_weeks = np.split(two_weeks, 2, axis=1)

print("\nFirst Week Matches:", np.array_equal(split_weeks[0], week1))
print("Second Week Matches:", np.array_equal(split_weeks[1], week2))

# Flatten and ravel comparison
flat_array = two_weeks.flatten()
ravel_array = two_weeks.ravel()

print("\nFlatten Shape:", flat_array.shape)
print("Ravel Shape:", ravel_array.shape)

print("Flatten Shares Memory:", np.shares_memory(two_weeks, flat_array))
print("Ravel Shares Memory:", np.shares_memory(two_weeks, ravel_array))
```

## Task 3: Daily Revenue and Item Co-Sales Matrix

Regenerate Session 01 data using seed 42. Calculate daily revenue using dot(), matmul() and element-wise multiplication. Verify all three results and build the item co-sales matrix.

```python
# Regenerate Session 01 data exactly
np.random.seed(42)
units_sold = np.random.randint(5, 60, size=(6, 7))
price_inr = np.array([52, 44, 45, 380, 140, 120])

print("Units Sold:")
print(units_sold)

# Method 1: Dot product
daily_revenue_1 = price_inr.dot(units_sold)

# Method 2: Matrix multiplication
daily_revenue_2 = np.matmul(price_inr, units_sold)

# Method 3: Element-wise multiplication and sum
daily_revenue_3 = (units_sold * price_inr.reshape(6, 1)).sum(axis=0)

print("\nRevenue using dot():", daily_revenue_1)
print("Revenue using matmul():", daily_revenue_2)
print("Revenue using element-wise multiplication:", daily_revenue_3)

# Verify all three methods
print("\nMethod 1 and 2 Match:", np.allclose(daily_revenue_1, daily_revenue_2))
print("Method 1 and 3 Match:", np.allclose(daily_revenue_1, daily_revenue_3))
print("Method 2 and 3 Match:", np.allclose(daily_revenue_2, daily_revenue_3))

# Item co-sales matrix
item_cosales = units_sold @ units_sold.T

print("\nItem Co-Sales Matrix:")
print(item_cosales)
print("\nCo-Sales Matrix Shape:", item_cosales.shape)
```

## Task 4: Weekly Statistical Summary

Calculate total, mean, median, standard deviation and variance for each item. Identify the day with the highest total units sold using NumPy only.

```python
item_names = np.array(["Rice", "Atta", "Sugar", "Tea", "Oil", "Dal"])

# Item-wise statistical calculations
item_total = units_sold.sum(axis=1)
item_mean = units_sold.mean(axis=1)
item_median = np.median(units_sold, axis=1)
item_std = np.std(units_sold, axis=1, ddof=0)
item_variance = np.var(units_sold, axis=1)

# Create summary table using NumPy
weekly_summary = np.column_stack((
    item_names,
    item_total,
    item_mean,
    item_median,
    item_std,
    item_variance
))

print("Weekly Summary:")
print("Item | Total | Mean | Median | Std | Variance")
print(weekly_summary)

# Daily totals and highest sales day
daily_totals = units_sold.sum(axis=0)
highest_sales_day = np.argmax(daily_totals) + 1

print("\nDaily Total Units:")
print(daily_totals)

print("\nHighest Sales Day:", highest_sales_day)
print("Highest Daily Units:", daily_totals[highest_sales_day - 1])
```
