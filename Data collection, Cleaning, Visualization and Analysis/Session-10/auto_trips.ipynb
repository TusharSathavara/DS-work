# Session 10 – Data Visualization: Matplotlib (Part 1)

This notebook generates 600 fictional auto-rickshaw trips and creates four Matplotlib visualizations:
1. Daily total fare line plot
2. Fare vs distance scatter plot
3. Trips per area bar chart
4. Fare distribution histogram


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# Reproducibility
np.random.seed(42)
rng = np.random.default_rng(42)

# Task 1: Generate 600 fictional auto-rickshaw trips
n = 600
start_date = pd.Timestamp("2025-04-01")
trip_dates = start_date + pd.to_timedelta(rng.integers(0, 30, size=n), unit="D")

areas = ["Navrangpura", "Maninagar", "Satellite", "Vastrapur", "Paldi"]
distance = rng.uniform(1, 15, size=n).round(2)
noise = rng.uniform(-15, 25, size=n)
fare = np.round(30 + 14 * distance + noise).astype(int)

trips = pd.DataFrame({
    "trip_id": [f"TRIP{i:04d}" for i in range(1, n + 1)],
    "trip_date": trip_dates,
    "area": rng.choice(areas, size=n),
    "distance_km": distance,
    "fare_inr": fare
})

trips.to_csv("auto_trips.csv", index=False)
print("Dataset shape:", trips.shape)
trips.head()
```

```python
# Reload the saved dataset and ensure trip_date is datetime
trips = pd.read_csv("auto_trips.csv", parse_dates=["trip_date"])
print(trips.dtypes)
```

## Task 1 – Daily total fare line plot

```python
daily_fare = trips.set_index("trip_date")["fare_inr"].resample("D").sum()

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(daily_fare.index, daily_fare.values, marker="o", linewidth=1.8, label="Daily total fare")
ax.set_title("Daily Total Auto-Rickshaw Fare")
ax.set_xlabel("Trip Date")
ax.set_ylabel("Total Fare (INR)")
ax.legend()
ax.grid(True, alpha=0.25)
fig.autofmt_xdate()
fig.tight_layout()
fig.savefig("daily_fare_line.png", dpi=160)
plt.show()
```

## Task 2 – Fare vs distance scatter chart

```python
short_trips = trips[trips["distance_km"] <= 5]
long_trips = trips[trips["distance_km"] > 5]

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(short_trips["distance_km"], short_trips["fare_inr"],
           color="orange", alpha=0.6, label="Trips up to 5 km")
ax.scatter(long_trips["distance_km"], long_trips["fare_inr"],
           color="teal", alpha=0.6, label="Trips over 5 km")
ax.set_title("Auto-Rickshaw Fare vs Distance")
ax.set_xlabel("Distance (km)")
ax.set_ylabel("Fare (INR)")
ax.legend()
fig.tight_layout()
fig.savefig("fare_vs_distance.png", dpi=160)
plt.show()
```

## Task 3 – Trips per area bar chart

```python
area_counts = trips["area"].value_counts().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(8, 5))
bar_colors = plt.cm.Set2(np.linspace(0, 1, len(area_counts)))
ax.bar(area_counts.index, area_counts.values, color=bar_colors)
ax.set_title("Number of Auto-Rickshaw Trips per Area")
ax.set_xlabel("Area")
ax.set_ylabel("Number of Trips")
ax.tick_params(axis="x", rotation=20)
fig.tight_layout()
fig.savefig("trips_per_area.png", dpi=160)
plt.show()
```

## Task 4 – Fare histogram and mean fare

```python
mean_fare = trips["fare_inr"].mean()

fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(trips["fare_inr"], bins=15, edgecolor="black",
        alpha=0.75, label="Fare distribution")
ax.axvline(mean_fare, linestyle="--", linewidth=2,
           label=f"Mean fare = ₹{mean_fare:.2f}")
ax.set_title("Distribution of Auto-Rickshaw Fares")
ax.set_xlabel("Fare (INR)")
ax.set_ylabel("Frequency")
ax.legend()
fig.tight_layout()
fig.savefig("fare_hist.png", dpi=160)
plt.show()

# Confirm all four image files were saved
for filename in ["daily_fare_line.png", "fare_vs_distance.png",
                 "trips_per_area.png", "fare_hist.png"]:
    print(filename, Path(filename).exists())
```
